```python
import numpy as np

import os
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

```

```python


import re

def load_images_from_folder(folder):
    images = []
    labels = []
    print(f"Loading images from folder: {folder}")

    if not os.path.exists(folder):
        print(f"Error: Folder {folder} does not exist.")
        return np.array([]), np.array([])

    filenames = os.listdir(folder)
    print("Filenames:", filenames)

    for filename in filenames:
        file_path = os.path.join(folder, filename)
        print(f"Processing file: {file_path}")

        if not os.path.isfile(file_path):
            print(f"Skipping {file_path}, not a file.")
            continue

        img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            print(f"Failed to load image: {file_path}")
            continue

        img = cv2.resize(img, (64, 64))  # Resize image to 64x64
        images.append(img.flatten())  # Flatten the image

        # Extract the integer label using regular expression
        match = re.match(r'image_(\d+)', filename)
        if match:
            try:
                label = int(match.group(1))
                labels.append(label)
            except ValueError:
                print(f"Failed to convert label to integer from filename: {filename}")
                continue
        else:
            print(f"No label found in filename: {filename}")
            continue

    print(f"Loaded {len(images)} images.")
    return np.array(images), np.array(labels)

# Correctly specify the folder path
folder = r'C:\Users\username\Documents\path_to_dataset'  # Example using raw string
X, y = load_images_from_folder(folder)

# Print labels to verify
print("Images", X)

```
