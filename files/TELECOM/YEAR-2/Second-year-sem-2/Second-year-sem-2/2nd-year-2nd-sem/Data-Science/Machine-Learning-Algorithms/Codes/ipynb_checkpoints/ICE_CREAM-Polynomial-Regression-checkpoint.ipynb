```python
#Importing the necessary libraries necessary for the data modeling.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.preprocessing import StandardScaler
```

```python
# Load the dataset
data = pd.read_csv('Ice_cream selling data.csv')

# Display the first few rows of the dataset
#data.head()
print(data)
```

```python
# Step 2: Data Preprocessing

X = data['Temperature']
y = data['Ice Cream Sales']
```

```python
# Visualize the data by plotting 
plt.plot(data['Temperature'], data['Ice Cream Sales'], 'b.')
plt.xlabel("Temperature")
plt.ylabel("Revenue")
plt.show()

X = data[['Temperature']].values
y = data['Ice Cream Sales'].values.reshape(-1, 1)
```

```python
model = LinearRegression()
model.fit(X,y)
y_pred = model.predict(X)

plt.plot(data['Temperature'], data['Ice Cream Sales'], 'b.', label='Data Points')
plt.plot(data['Temperature'], y_pred, 'r-', label='Regression Line')
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
plt.title("Temperature vs Ice Cream Sales with Regression Line")
plt.legend()
plt.show()
```

```python
#TSplitting the data
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)

# Applying polynomial regression with degree 5
poly = PolynomialFeatures(degree=5, include_bias=True)
x_train_trans = poly.fit_transform(x_train)
x_test_trans = poly.transform(x_test)
```

```python
#Training the model with data
lr = LinearRegression()
lr.fit(x_train_trans, y_train)
y_pred = lr.predict(x_test_trans)
```

```python
# Visualize the polynomial regression
X_new = np.linspace(min(X), max(X), 200).reshape(200, 1)
X_new_poly = poly.transform(X_new)
y_new = lr.predict(X_new_poly)
```

```python
#Fitting your p
plt.plot(X_new, y_new, "r-", linewidth=2, label="Predictions")
plt.plot(x_train, y_train, "b.", label='Training points')
plt.plot(x_test, y_test, "g.", label='Testing points')
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Sales")
plt.legend()
plt.show()
```

```python
#Generating the coefficients
print("R2 Score for Polynomial Regression (degree=5):", r2_score(y_test, y_pred))
print("Coefficients:", lr.coef_)
print("Intercept:", lr.intercept_)
```

```python
#Making predictions using your model
pred2 = 12
pred2array = np.array([[pred2]])
lr.predict(poly.fit_transform(pred2array))
```

```python
#Calculating for the errors generated 
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse}")
print(f"Root Mean Squared Error (RMSE): {rmse}")
print(f"R-squared (R²): {r2}")
```
