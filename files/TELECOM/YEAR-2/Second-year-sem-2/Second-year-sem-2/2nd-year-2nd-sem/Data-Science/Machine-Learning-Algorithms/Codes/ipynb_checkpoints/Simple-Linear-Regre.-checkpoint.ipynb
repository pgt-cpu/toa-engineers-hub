```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
square_footage = np.array([1500,1600,1700,1800,1900,2000])
house_prices = np.array([300000,320000,340000,360000,380000,400000])
```

```python
#Plotting and Visualizing
plt.scatter(square_footage,house_prices,color='red')
plt.xlabel('square footage')
plt.ylabel('house prices')
plt.title('House price vs Square footage')
plt.show()
```

```python
#Import Linear Regression Model
from sklearn.linear_model import LinearRegression
X= square_footage.reshape(-1,1)
Y= house_prices

model = LinearRegression() #creating the linear regression model
model.fit(X,Y) #fitting the model
intercept= model.intercept_ #intercept
slope= model.coef_ [0]#slope
print(f"Intercept: {intercept}")
print(f"Slope:{slope}")
```

```python
#Evaluating the Model
from sklearn.metrics import mean_squared_error,r2_score
Y_pred = model.predict(X)
mse= mean_squared_error(Y,Y_pred)
r2= r2_score(Y,Y_pred)
print(f"MSE: {mse}")
print(f"R2: {r2}")
```

```python
#Plotting
plt.scatter(X, Y, color='blue')
plt.plot(X, Y_pred, color='red')
plt.xlabel('square footage')
plt.ylabel('House prices')
plt.title('Simple Linear Regression')
plt.show()
```

```python
#Making New Predictions
new_square_footage = np.array([2100]).reshape(-1,1)
predicted_price = model.predict(new_square_footage)
print(f"Predicted price for a house with {new_square_footage.flatten()} :{predicted_price}")
```
