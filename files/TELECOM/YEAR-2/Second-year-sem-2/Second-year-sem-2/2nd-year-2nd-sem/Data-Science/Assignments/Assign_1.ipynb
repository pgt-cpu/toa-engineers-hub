```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_excel("sales_data.xlsx")
```

```python
df.head()
```

```python
## creating a line graph to compare sales by month
x_axis = df['Month '] 
y_axis = df['Sales Amount ']
plt.xlabel('Month')
plt.ylabel('Sales Amount ')
plt.plot(x_axis, y_axis)
plt.title('Camparing Sales Amount by Month')
plt.show
```

```python
## creating a bar graph to compare sales by product category
x_axis = df['Product Category'] 
y_axis = df['Sales Amount ']
plt.bar(x_axis, y_axis)
plt.xlabel('Product Category')
plt.ylabel('Sales Amount ')
plt.title('Camparing Sales by Product Category')
plt.show()
```

```python
##visualizing sales distribution Across Regions and Months
sales_pivot = df.pivot_table(index='Region ', columns='Month ', values='Sales Amount ', aggfunc='sum')
plt.figure(figsize=(10, 5))
sns.heatmap(sales_pivot, cmap='plasma', annot=True, fmt='.0f')
plt.title('Sales Distribution Across Regions And Months')
plt.xlabel('MONTHS')
plt.ylabel('REGIONS')
plt.xticks(rotation=45)
plt.show()
```
